```python
import numpy as np

print("NumPy version:", np.__version__)
```

Scalar

```python
x = np.array(5)
print(x)
print("shape: ", x.shape)
print('dimensions: ', x.ndim)
```

Vector

```python
x = np.array([1, 2, 3, 4])
print(x)
print("shape: ", x.shape)
print("dimensions: ", x.ndim)
```

```python
x = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(x)
print('shape: ', x.shape)
print('dimensions: ', x.ndim)
```

```python
x = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])
print(x)
print('shape: ', x.shape)
print('dimensions: ', x.ndim)
```

```python
x = np.array([
    [10,20,30],
    [40,50,60]
])
print(x[0,0])
print(x[0,1])
print(x[1,2])
```

```python
x = np.array([1, 2, 3, 4, 5])
print(x * 10)
print(x + 10)
print(x ** 2)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
```

```python
C = A @ B
print(C)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
print("A shape:", A.shape)
print("B shape:", B.shape)
print("A @ B shape:", (A @ B).shape)
```
